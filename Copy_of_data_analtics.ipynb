```python
import pandas as pd
data = {
    "Name":["Ravi", "priya", "Arun", "Sneha", "kiran", "Anjali", "Vikram", "Meena"],
    "Age":[22, 24, None, 25, None, 26, 20, 27],
    "city":["Hyderabad", "Chennai", "Bangalore", "Hyderabad",
            "Mumbai", None, "Bangalore", "Mumbai"],
    "Salary":[30000, 40000, 28000, 50000, 35000, 45000, None, 55000]

}
df = pd.DataFrame(data)
print(df)
df.iloc[0]
print("_____________________________________")
e = df["Age"]>21
print(e)
print("_____________________________________")
d=df[df["Age"]>21]
print(d)
#df-boolean values which is true or false
#df[df["f"]]
f=df[df["Salary"]>40000]
print(f)
print("_____________________________________")
g=df[df["city"]=="Hyderabad"]
print(g)
k=df[(df["Age"]>22) &(df["city"]=='Hyderabad')]
print(k)
print("_____________________________________")
h=df[
    (df["Salary"]>45000)|
 (df["city"]=='Mumbai')
 ]
print(h)
df.isna()
df.isna().sum()
df.dropna()
df.dropna(axis=0)
df.dropna(how='all')
print("_____________________________________")
df.dropna(axis=0,inplace=True)
print(df)
#filling with different kinds of values
df.fillna({
    'Age':0,
    'Salary':'700000'
})
print(df)
df['Age'] == df['Age'].fillna(
    df['Age'].mean())





```

```python
import pandas as pd
emp_details = {
    "EmoID": [101, 102, 103, 104, 101],
    "City": ["Hyderabad", "Chennai", "Bangalore", None, "Hyderabad"],
    "Experience": [25, 30, None, 35, 25],
    "age":[25, 30, None, 35, 25]

}

df = pd.DataFrame(emp_details)
print(df)
df.isna()
df.isna().sum()
print
df.fillna({'City':'Unkown','Experience':0})
print(df)
df['age'] = df['age'].fillna(
    df['age'].median())
print(df)

df["salary"]= [1000, 2000, 3000, 4000, None]
print(df)
df["Bonus"]= df["salary"]*10/100
print(df)
df.drop(
    'Bonus',
    axis=1,
    inplace=True)
print(df)
#for multiple values
df.drop(
    columns=['salary'],
    inplace=True
)
df.loc[len(df)] = [109,'pune',4,25]
print(df)
df.drop(2,
        axis=0,
        inplace=True
        )
print(df)
#basix aggrigation & groupby &orderby
```

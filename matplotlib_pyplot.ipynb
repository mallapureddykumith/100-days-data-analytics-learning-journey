```python

#line
import matplotlib.pyplot as plt
import numpy as np
products =["Laptop","Mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.plot(products,sales,marker="^")
plt.title("products and sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(products,sales,marker="^",label="sales")
plt.plot(products,profit,marker="*",label="loss")
plt.title("sales and loss")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
#bar
plt.bar(products,sales,label="sales")
plt.title("product sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#multiple/cluster bar chart
x= np.arange(len(products))
width=0.35
plt.bar(x-width/2,sales,width,label="sales")
plt.bar(x+width/2,profit,width,label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
#stacked chart
x= np.arange(len(products))
width=0.35
plt.bar(x+width/2,sales,width,label="sales")
plt.bar(x+width/2,profit,width,label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
import pandas as pd
#dataframe creation
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Math": [65, 70, 75, 80, 85, 60, 90, 72, 78, 88],
    "Python": [70, 75, 80, 85, 90, 65, 95, 78, 82, 92],
    "SQL": [60, 68, 72, 78, 82, 55, 88, 70, 75, 85]
}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
#violine plot-math marks distribution
plt.violinplot(df["Math"])
plt.title("Distribution of Math Marks")
plt.ylabel("Marks")
plt.show()
```

```python
#stem plot
plt.stem(df["Student"], df["Math"])
plt.title("Student Math Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#countour plot
import numpy as np
import matplotlib.pyplot as plt

x = [1,2,3]
y = [10,20]
X, Y = np.meshgrid(x, y)
Z = X*2 + Y*2

plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#heatmap
marks = df[["Math", "Python", "SQL"]].values
plt.imshow(marks)
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar()
```

```python
marks = df[["Math", "Python", "SQL"]].values
plt.imshow(marks, cmap="Reds")
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar()
```

```python
import matplotlib.pyplot as plt
# 3D Scatter Plot
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
z = [5, 10, 8, 15, 20]

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.scatter(x, y, z)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
import matplotlib.pyplot as plt

# 3D Line Plot
x = [1, 2, 3, 4, 5]  # Month
y = [100, 150, 120, 180, 200]  # Sales
z = [20, 30, 25, 40, 50]  # Profit

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.plot(x, y, z)

ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.set_zlabel("Profit")
ax.set_title("3D Sales and Profit")

plt.show()
```

```python
import matplotlib.pyplot as plt

# 3D Bar Chart
x = [1, 2, 3, 4]
y = [1, 2, 3, 4]
z = [0, 0, 0, 0]

dx = [0.5, 0.5, 0.5, 0.5]  # width
dy = [0.5, 0.5, 0.5, 0.5]  # depth
dz = [10, 15, 10, 20]      # height

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.bar3d(x, y, z, dx, dy, dz)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 3D Wireframe Plot
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)

X, Y = np.meshgrid(x, y)
Z = X*2 + Y*2

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.plot_wireframe(X, Y, Z)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.barplot(data=data, x="day", y="total_bill")
plt.title("Barplot")
plt.show()

```

```python
sns.scatterplot(data=data, x="total_bill", y="tip")
plt.title("Scatterplot")
plt.show()

```

```python
sns.lineplot(data=data, x="size", y="tip")
plt.title("Lineplot")
plt.show()

```

```python
sns.boxplot(data=data, x="day", y="total_bill")
plt.title("Boxplot")
plt.show()

```

```python
sns.violinplot(data=data, x="sex", y="tip")
plt.title("Violinplot")
plt.show()
```

```python
import matplotlib.pyplot as plt
sns.histplot(data=data, x="total_bill", bins=20)
plt.title("Histplot")
plt.show()
```
